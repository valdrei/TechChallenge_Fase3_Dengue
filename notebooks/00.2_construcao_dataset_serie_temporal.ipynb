```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import matplotlib.gridspec as gridspec
import warnings

warnings.filterwarnings("ignore")
```

```python
# Carregamento dos dados
df_bruto = pd.read_csv(r'../data/raw/dados_dengue_clima_saneamento_2014_2025.csv')
```

```python
df_bruto.head()
```

```python
# Criação do DataFrame de série temporal
df_ts = df_bruto[['periodo', 'Quantidade de Casos', 'COD_UF']].copy()
```

```python
print('Informações básicas do DataFrame\n')
df_ts.info()

print('Verificando o Conteúdo do DataFrame\n')
df_ts.head()
```

```python
print('Verificando valores ausentes\n')
df_ts.isna().sum()
```

```python
# Renomeando colunas para facilitar o uso com modelos de séries temporais
df_ts.rename(columns={'Quantidade de Casos': 'y', 'COD_UF': 'estado'}, inplace=True)

# Convertendo a coluna 'periodo' para o tipo datetime e depois para Period
df_ts['periodo'] = pd.to_datetime(df_ts['periodo'], format='%Y-%m').dt.to_period('M')

df_ts.info()
```

```python
# Salvar o dataframe df_ts em um arquivo CSV
caminho_salvar = r'../data/raw/dataset_serie_temporal.csv'
df_ts.to_csv(caminho_salvar, index=False)
print(f"DataFrame df_ts salvo com sucesso em: {caminho_salvar}")
```
